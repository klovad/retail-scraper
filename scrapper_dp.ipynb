```python
import pandas as pd
import numpy as np
import unicodedata
import os
import csv

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import time
```

```python
folders = ['data/raw/', 'data/processed/', 'data/final/']

for folder in folders:
    if not os.path.exists(folder):
        os.makedirs(folder)
        print(f"Carpeta creada: {folder}")
    else:
        print(f"La carpeta ya existe: {folder}")
```

```python
def eliminar_tildes(cadena):
    s = ''.join((c for c in unicodedata.normalize('NFD',cadena) if unicodedata.category(c) != 'Mn'))
    return s
```

```python
# ruta del driver de chromium para el scraper
driver = r"C:\Users\User\Documents\proyectos\chromedriver-win64\chromedriver.exe"

# LLama al driver de Chromium
service = Service(executable_path=driver)
driver = webdriver.Chrome(service=service)

# Abre el navegador chrome en la pagina principal de la super de bancos
driver.get("https://www.deprati.com.ec/")

xpath_seccion_mujeres = "//div[@data-test-btn-menu-item='btn_menu_Mujeres']/a"
# Da tiempo para que cargue la pagina
WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.XPATH, xpath_seccion_mujeres))
)

# Apunta al icono con el link de calificacion de riesgo
navegador_seccion_mujeres = driver.find_element(By.XPATH, xpath_seccion_mujeres)
# Click ingreso a seccion MUJERES
navegador_seccion_mujeres.click()

# Mujeres - vestidos
xpath_mujeres_vestidos = "//a[@href='https://www.deprati.com.ec/es/vestidos-y-enterizos/c/0123']"
# Da tiempo para que cargue la pagina
WebDriverWait(driver, 5).until(
    EC.presence_of_element_located((By.XPATH, xpath_mujeres_vestidos))
)

# Apunta al icono con el link de calificacion de riesgo
navegador_seccion_mujeres_vestidos = driver.find_element(By.XPATH, xpath_mujeres_vestidos)
# Click ingreso a seccion MUJERES
navegador_seccion_mujeres_vestidos.click()

# Extrae la descripcion de los productos visibles
productos = driver.find_elements(By.CLASS_NAME, "info--table")

# Para quemar el titulo de la categoria sobre el cual se esta descargando los porductos
xpath_titulo_categoria =  "//div[@class='product--grid--page__page-title']/h1"
titulo_categoria = eliminar_tildes(driver.find_element(By.XPATH, xpath_titulo_categoria).text.strip())

# Abre el archivo CSV y prepara el escritor
with open(f'{folders[0]}productos_adp.csv', mode='w', newline='', encoding='utf-8') as archivo_csv:
    escritor = csv.writer(archivo_csv, delimiter="^")
    
    # Escribe el encabezado
    escritor.writerow(['nombre', 'precio_regular', 'precio_venta', 'empresa', 'categoria'])

    # Iterar sobre cada producto
    for producto in productos:
        try:
            nombre = eliminar_tildes(producto.find_element(By.XPATH, ".//div[@class='title--cell']/a").text.strip())
        except:
            nombre = "N/A"
        
        try:
            precio_regular = producto.find_element(By.XPATH, ".//div[@class='price--cell']/div[@class='price']").text.strip()
        except:
            precio_regular = "N/A"
        
        try:
            precio_descuento = producto.find_element(By.XPATH, ".//div[@class='price--cell']/div[@class='disccount']").text.strip()
        except:
            precio_descuento = "N/A"
        
        # Escribe la fila al CSV
        escritor.writerow([nombre, precio_regular, precio_descuento, 'ADP', titulo_categoria])
        print(f"Nombre: {nombre} | Precio regular: {precio_regular} | Descuento: {precio_descuento}")


time.sleep(10)

driver.quit()
```
